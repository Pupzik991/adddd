```python
import numpy as np
from sklearn.linear_model import LinearRegression


X_train = np.array([[-2, 3], [1, 3]])
Y_train = np.array([
    [60, 40, 100],
    [200, 50, 80]
])


model = LinearRegression(fit_intercept=False)
model.fit(X_train, Y_train)


raw_prediction = model.predict([[-7, 6]])[0]
print("Сырое предсказание сети:", raw_prediction)

final_color = np.clip(raw_prediction, 0, 255)
print("Итоговый цвет пикселя:", final_color)
```

```python
import numpy as np
from sklearn.linear_model import QuantileRegressor

points = [
    (0,0), (0,2), (0,7), (0,8),
    (4,4), (4,7),
    (6,1), (6,5),
    (9,0), (9,1), (9,4), (9,7),
    (11,2), (11,6)
]

X = np.array([[p[0]] for p in points])
y = np.array([p[1] for p in points])

model = QuantileRegressor(quantile=0.5, alpha=0)


model.fit(X, y)

y_pred = model.predict(X)

mae = np.mean(np.abs(y_pred - y))
print(mae)
```

```python
import numpy as np
from sklearn.svm import SVC

input_data = """
0.0 0.0 -1
0.5 0.5 -1
-0.5 0.5 -1
2.0 2.0 1
-2.0 2.0 1
2.0 -2.0 1
-2.0 -2.0 1
"""

lines = input_data.strip().split('\n')
data = np.array([list(map(float, line.split())) for line in lines if line.strip()])

X_2d = data[:, :2]
y = data[:, 2]

z = X_2d[:, 0]**2 + X_2d[:, 1]**2
X_3d = np.column_stack((X_2d, z))


model = SVC(kernel='linear')
model.fit(X_3d, y)


w1, w2, w3 = model.coef_[0]
b = model.intercept_[0]


x0 = -w1 / (2 * w3)
y0 = -w2 / (2 * w3)
R = np.sqrt((w1**2 + w2**2 - 4 * w3 * b) / (4 * w3**2))

print(f"Центр окружности (x0, y0): {x0:.4f}, {y0:.4f}")
print(f"Радиус R: {R:.4f}")
```
